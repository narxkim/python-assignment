```python
import numpy as np, pandas as pd, matplotlib.pyplot as plt
import platform

from sklearn.cluster import KMeans
from sklearn.decomposition import PCA

# 한글 폰트 설정
if platform.system() == "Windows":
    plt.rcParams["font.family"] = "Malgun Gothic"
elif platform.system() == "Darwin":
    plt.rcParams["font.family"] = "AppleGothic"
else:
    plt.rcParams["font.family"] = "NanumGothic"

# 마이너스 깨짐 방지
plt.rcParams["axes.unicode_minus"] = False
```

```python
fruits = np.load('./fruits_300.npy')
fruits.shape
```

```python
fruits_2d = fruits.reshape(-1, 100*100)
fruits_2d.shape
```

```python
km = KMeans(n_clusters=3, random_state=42, n_init=10)

km.fit(fruits_2d)

print(np.unique(km.labels_, return_counts=True))
print(km.n_iter_)
```

```python
km.cluster_centers_
```

```python
km.inertia_
```

```python
def draw_fruits(arr, ratio=1):
    n = len(arr)
    rows = int(np.ceil(n / 10))
    cols = n if rows < 2 else 10
    fig, axs = plt.subplots(
        rows, cols, figsize=(cols * ratio, rows * ratio), squeeze=False
    )
    for i in range(rows):
        for j in range(cols):
            if i * 10 + j < n:
                axs[i, j].imshow(arr[i * 10 + j], cmap="gray_r")
            axs[i, j].axis("off")
    plt.show()
```

```python
draw_fruits(fruits[km.labels_ == 0]) # pineapple
```

```python
draw_fruits(fruits[km.labels_ == 1]) # banana
```

```python
draw_fruits(fruits[km.labels_ == 2]) # apple
```

```python
#  KMeans centroid 이미지
draw_fruits(km.cluster_centers_.reshape(-1, 100, 100), ratio=3)
```

```python
# 실제 과일별 평균 이미지
apple_mean = fruits[0:100].mean(axis=0)
pineapple_mean = fruits[100:200].mean(axis=0)
banana_mean = fruits[200:300].mean(axis=0)
```

```python
# centroid 이미지는 KMeans가 정답 없이 찾은 군집 중심이고, 평균 이미지는 실제 과일 순서를 알고 계산한 기준 이미지이다.
draw_fruits(np.array([pineapple_mean, banana_mean, apple_mean]), ratio=3)
```

## PCA

```python
pca = PCA(n_components=50)

fruits_pca = pca.fit_transform(fruits_2d)

print("원본 데이터:", fruits_2d.shape)
print("PCA 변환 후:", fruits_pca.shape)
```

```python
print("설명력 비율")
print(pca.explained_variance_ratio_)

print("\n누적 설명력")
print(np.sum(pca.explained_variance_ratio_))
```

```python
plt.figure(figsize=(8, 5))

plt.plot(np.cumsum(pca.explained_variance_ratio_), marker="o")

plt.xlabel("주성분 개수")
plt.ylabel("누적 설명력 비율")
plt.title("PCA 누적 설명력")

plt.grid(True)

plt.show()
```

```python
# 2차원 시각화용 PCA
pca2 = PCA(n_components=2)

fruits_pca2 = pca2.fit_transform(fruits_2d)

print("PCA 2D shape:", fruits_pca2.shape)
```

```python
# PCA 2D 결과를 KMeans 군집 라벨로 색칠
plt.figure(figsize=(8, 6))

for label in range(3):
    data = fruits_pca2[km.labels_ == label]

    plt.scatter(data[:, 0], data[:, 1], label=f"군집 {label}", alpha=0.8)

plt.xlabel("주성분 1")
plt.ylabel("주성분 2")
plt.title("과일 데이터 PCA 2차원 시각화")
plt.legend()
plt.grid(True)

plt.show()
```

###  LR 학습시간 비교 표

#### 1. 라이브러리 불러오기 및 정답 라벨 생성

PCA 적용 전후의 학습 시간을 비교하기 위해 `time`, `pandas`, `LogisticRegression`을 불러온다.

과일 데이터는 순서대로 저장되어 있으므로 직접 정답 라벨(y)을 생성한다.

- 0 ~ 99 : Apple
- 100 ~ 199 : Pineapple
- 200 ~ 299 : Banana

총 300개의 이미지에 대해 3개의 클래스를 구성한다.



```python
import time
import pandas as pd
from sklearn.linear_model import LogisticRegression

# 실제 과일 순서 기준 라벨
# 0~99: apple, 100~199: pineapple, 200~299: banana
y = np.array([0] * 100 + [1] * 100 + [2] * 100)
```


---

#### 2. 원본 데이터 학습 시간 측정

원본 데이터(`fruits_2d`)는 100 × 100 이미지를 펼친 10,000개의 특성을 가진다.

`time.time()`을 사용하여 학습 시작 시각을 저장한 뒤,
`LogisticRegression` 모델을 학습시키고,
학습 종료 시각과의 차이를 계산하여 학습 시간을 측정한다.

또한 `score()`를 사용하여 원본 데이터에 대한 분류 정확도를 확인한다.

입력 크기:
- (300, 10000)

목적:
- PCA 적용 전 기준(Baseline) 성능 측정



```python
# 원본 10000차원 데이터 학습시간
lr_origin = LogisticRegression(max_iter=1000)

start = time.time()
lr_origin.fit(fruits_2d, y)
origin_time = time.time() - start

origin_score = lr_origin.score(fruits_2d, y)
```


---
## 3. PCA 데이터 학습 시간 측정

PCA를 통해 10,000개의 픽셀 특성을 50개의 주성분으로 압축한 데이터를 사용한다.

동일한 Logistic Regression 모델을 학습시키고,
원본 데이터와 동일한 방식으로 학습 시간을 측정한다.

입력 크기:
- (300, 50)

목적:
- 차원 축소가 학습 속도에 미치는 영향 확인
- 정보 손실 없이 계산량을 줄일 수 있는지 검증

```python
# PCA 50차원 데이터 학습시간
lr_pca = LogisticRegression(max_iter=1000)

start = time.time()
lr_pca.fit(fruits_pca, y)
pca_time = time.time() - start

pca_score = lr_pca.score(fruits_pca, y)
```


---

#### 4. 학습 시간 및 정확도 비교

원본 데이터와 PCA 데이터의 결과를 하나의 DataFrame으로 정리한다.

비교 항목:
- 데이터 종류
- 입력 데이터 크기
- 학습 시간(초)
- 정확도

분석 포인트:
- PCA 적용 시 학습 시간이 얼마나 감소했는가?
- 차원 축소 후에도 정확도가 유지되는가?

PCA는 고차원 데이터를 저차원으로 압축하여
계산 비용을 줄이고 학습 속도를 향상시키는 대표적인 차원 축소 기법이다.

```python
# 학습시간 비교 표
time_df = pd.DataFrame(
    {
        "데이터": ["원본 데이터", "PCA 데이터"],
        "입력 크기": [str(fruits_2d.shape), str(fruits_pca.shape)],
        "학습시간(초)": [origin_time, pca_time],
        "정확도": [origin_score, pca_score],
    }
)

time_df
```

#### 결과 해석

원본 데이터는 10,000개의 특성을 사용하여 학습하므로 계산량이 많다.

반면 PCA는 데이터를 50개의 주성분으로 압축하여
중요한 정보는 유지하면서 불필요한 중복 정보를 제거한다.

따라서 학습 시간이 크게 감소하며,
설명력이 충분히 유지된다면 정확도 역시 큰 손실 없이 유지될 수 있다.

즉, PCA는 고차원 데이터의 학습 효율을 향상시키는 효과적인 전처리 기법이다.
